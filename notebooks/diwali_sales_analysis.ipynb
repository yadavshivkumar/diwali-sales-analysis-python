###  State-wise Sales Analysis

#### Objective
This analysis evaluates sales performance across different states using the purchase amount recorded in the dataset.

#### Analysis Approach
- Group the dataset by state.
- Calculate total purchase amount for each state.
- Sort the results in descending order.
- Visualize the top 10 states by total purchase amount.

###  Occupation-wise Purchase Amount Analysis

#### Objective
This section evaluates the total recorded purchase amount associated with each occupation.

#### Methodology
- Group records by occupation.
- Sum the purchase amounts for each group.
- Sort the results in descending order.
- Visualize the top 10 occupations by total purchase amount.

#### Business Relevance
The analysis helps identify occupational groups that contribute the largest combined purchase amounts in the available dataset.

#### Limitation
Total purchase amount can be influenced by the number of records in each group. It does not directly measure individual customer purchasing power or profitability.

```python
# Group the dataset by occupation and sum purchase amounts
occupation_sales = df.groupby("Occupation")["Amount"].sum()

# Sort occupations from highest to lowest purchase amount
occupation_sales = occupation_sales.sort_values(ascending=False)

# Display the top 10 occupations by total purchase amount
print("Top 10 occupations by total purchase amount:")
print(occupation_sales.head(10))
```

```python
# Select the top 10 occupations by total purchase amount
top_10_occupation_sales = occupation_sales.head(10)

# Create a figure with sufficient space for labels
plt.figure(figsize=(12, 6))

# Draw a bar chart of total purchase amount by occupation
sns.barplot(
    x=top_10_occupation_sales.index,
    y=top_10_occupation_sales.values
)

# Add a descriptive chart title
plt.title("Top 10 Occupations by Total Purchase Amount")

# Label the horizontal axis
plt.xlabel("Occupation")

# Label the vertical axis
plt.ylabel("Total Purchase Amount")

# Rotate labels for better readability
plt.xticks(rotation=45, ha="right")

# Adjust chart spacing
plt.tight_layout()

# Display the chart
plt.show()
```

###  Comparing Occupation-wise Records and Purchase Amounts

This table supports a clearer interpretation of customer representation and aggregate purchasing activity.

```python
# Create a summary table containing record counts by occupation
occupation_summary = occupation_counts.rename("Record_Count").to_frame()

# Add total purchase amount for each occupation
occupation_summary["Total_Amount"] = occupation_sales

# Sort the table by total purchase amount in descending order
occupation_summary = occupation_summary.sort_values(
    by="Total_Amount",
    ascending=False
)

# Display the top 10 occupations
display(occupation_summary.head(10))
```

###  Demographic Purchase Summary

- Number of records for each age group.
- Total purchase amount.
- Average purchase amount per record.

##  Customer Segmentation and Combined Insights

### Objective
The objective of this section is to combine customer demographic information with purchase data to identify important customer segments.

### Key Questions
- Which gender group has the highest total purchase amount?
- Which age group contributes the highest purchase amount?
- Which occupations have the highest purchase amounts?
- How do marital status groups compare?
- Which customer segments may be useful for targeted marketing?

### Business Purpose
Customer segmentation helps businesses understand different customer groups and develop more targeted marketing, promotional, and customer engagement strategies.

```python
# Group the dataset by gender
gender_summary = df.groupby("Gender").agg(
    # Count the number of records in each gender group
    Record_Count=("Gender", "size"),

    # Calculate the total purchase amount for each gender group
    Total_Amount=("Amount", "sum"),

    # Calculate the average purchase amount per record
    Average_Amount=("Amount", "mean")
)

# Sort gender groups by total purchase amount in descending order
gender_summary = gender_summary.sort_values(
    by="Total_Amount",
    ascending=False
)

# Display the summary table with rounded numeric values
display(gender_summary.round(2))
```

```python
# Group the dataset by age group
age_summary = df.groupby("Age Group").agg(
    # Count the number of records in each age group
    Record_Count=("Age Group", "size"),

    # Calculate total purchase amount for each age group
    Total_Amount=("Amount", "sum"),

    # Calculate average purchase amount per record
    Average_Amount=("Amount", "mean")
)

# Sort age groups by total purchase amount in descending order
age_summary = age_summary.sort_values(
    by="Total_Amount",
    ascending=False
)

# Display the age group summary
display(age_summary.round(2))
```

```python
# Group the dataset by occupation
occupation_summary = df.groupby("Occupation").agg(
    # Count the records for each occupation
    Record_Count=("Occupation", "size"),

    # Calculate total purchase amount for each occupation
    Total_Amount=("Amount", "sum"),

    # Calculate average purchase amount per record
    Average_Amount=("Amount", "mean")
)

# Sort occupations by total purchase amount in descending order
occupation_summary = occupation_summary.sort_values(
    by="Total_Amount",
    ascending=False
)

# Display the top 10 occupations by total purchase amount
display(occupation_summary.head(10).round(2))
```

```python
# Group the dataset by marital status
marital_summary = df.groupby("Marital_Status").agg(
    # Count records for each marital status code
    Record_Count=("Marital_Status", "size"),

    # Calculate total purchase amount for each group
    Total_Amount=("Amount", "sum"),

    # Calculate average purchase amount per record
    Average_Amount=("Amount", "mean")
)

# Sort groups by total purchase amount in descending order
marital_summary = marital_summary.sort_values(
    by="Total_Amount",
    ascending=False
)

# Display the marital status summary
display(marital_summary.round(2))
```

```python
# Identify the gender group with the highest total purchase amount
top_gender = gender_summary["Total_Amount"].idxmax()

# Identify the age group with the highest total purchase amount
top_age_group = age_summary["Total_Amount"].idxmax()

# Identify the occupation with the highest total purchase amount
top_occupation = occupation_summary["Total_Amount"].idxmax()

# Identify the marital status code with the highest total purchase amount
top_marital_status = marital_summary["Total_Amount"].idxmax()

# Display the highest-value groups based on total purchase amount
print("Top Segments by Total Purchase Amount")
print("--------------------------------------")
print("Top Gender Group:", top_gender)
print("Top Age Group:", top_age_group)
print("Top Occupation:", top_occupation)
print("Top Marital Status Code:", top_marital_status)
```
